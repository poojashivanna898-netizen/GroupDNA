```python
file_path = "/content/hostel_bois.txt"

with open(file_path, "r", encoding="utf-8") as file:
    lines = file.readlines()

print("Total lines:", len(lines))
print(lines[:5])
```

```python
messages = []

for line in lines:
    line = line.strip()

    # Skip empty lines
    if line == "":
        continue

    # Check if the line starts like a WhatsApp message
    if " - " not in line:
        continue

    date_time, message_part = line.split(" - ", 1)

    # Separate date and time
    if ", " not in date_time:
        continue

    date, time = date_time.split(", ", 1)

    # Separate sender and message
    if ": " not in message_part:
        continue

    sender, message = message_part.split(": ", 1)

    messages.append({
        "date": date,
        "time": time,
        "sender": sender,
        "message": message
    })

print("Parsed messages:", len(messages))
print(messages[:3])
```

```python
# Group Overview

total_messages = len(messages)

participants = set()

for msg in messages:
    participants.add(msg["sender"])

print("Total messages:", total_messages)
print("Participants:", participants)
print("Number of participants:", len(participants))
```

```python
# Message count for each participant

message_counts = {}

for msg in messages:
    sender = msg["sender"]

    if sender not in message_counts:
        message_counts[sender] = 0

    message_counts[sender] += 1

print("Messages per participant:")

for sender, count in message_counts.items():
    print(sender, ":", count)
```

```python
from datetime import datetime

dates = []

for msg in messages:
    date = datetime.strptime(msg["date"], "%d/%m/%y")
    dates.append(date)

start_date = min(dates)
end_date = max(dates)

number_of_days = (end_date - start_date).days + 1

print("Start date:", start_date.strftime("%d %B %Y"))
print("End date:", end_date.strftime("%d %B %Y"))
print("Number of days:", number_of_days)
```

```python
day_counts = {}

for msg in messages:
    date = msg["date"]

    if date not in day_counts:
        day_counts[date] = 0

    day_counts[date] += 1

busiest_day = max(day_counts, key=day_counts.get)

print("Busiest day:", busiest_day)
print("Messages on busiest day:", day_counts[busiest_day])


hour_counts = {}

for msg in messages:
    hour = int(msg["time"].split(":")[0])

    if hour not in hour_counts:
        hour_counts[hour] = 0

    hour_counts[hour] += 1

busiest_hour = max(hour_counts, key=hour_counts.get)

print("Busiest hour:", busiest_hour)
print("Messages during that hour:", hour_counts[busiest_hour])
```

```python
import numpy as np

participants = sorted(participants)

activity_heatmap = np.zeros((len(participants), 24), dtype=int)

for msg in messages:
    sender = msg["sender"]
    hour = int(msg["time"].split(":")[0])

    person_index = participants.index(sender)

    activity_heatmap[person_index][hour] += 1

print("Participants:", participants)
print("Heatmap shape:", activity_heatmap.shape)
print(activity_heatmap)
```

```python
# Printable Activity Heatmap

print("Activity Heatmap")
print("Each number represents the number of messages sent during that hour.")
print()

# Print hour headings
print("Participant", end="\t")

for hour in range(24):
    print(f"{hour:02}", end="\t")

print()

# Print each participant's activity
for i in range(len(participants)):
    print(participants[i], end="\t")

    for hour in range(24):
        print(activity_heatmap[i][hour], end="\t")

    print()
```

```python
# Top 10 most frequent words

stop_words = {
    "the", "is", "a", "an", "and", "or", "to", "of",
    "in", "on", "for", "with", "this", "that", "it",
    "i", "you", "we", "he", "she", "they", "was",
    "are", "be", "have", "has", "had", "my", "your",
    "our", "me", "but", "so", "at", "as", "from"
}

word_counts = {}

for msg in messages:
    text = msg["message"].lower()

    # Remove common punctuation
    punctuation = ".,!?;:'\"()[]{}<>-/\\"

    for character in punctuation:
        text = text.replace(character, " ")

    words = text.split()

    for word in words:
        if word == "" or word in stop_words:
            continue

        if word not in word_counts:
            word_counts[word] = 0

        word_counts[word] += 1

# Sort words by frequency
sorted_words = sorted(
    word_counts.items(),
    key=lambda item: item[1],
    reverse=True
)

print("Top 10 Words:")
for word, count in sorted_words[:10]:
    print(word, ":", count)
```

```python
# Convert messages into datetime values

for msg in messages:
    msg["datetime"] = datetime.strptime(
        msg["date"] + " " + msg["time"],
        "%d/%m/%y %H:%M"
    )

# Sort messages by date and time
messages.sort(key=lambda x: x["datetime"])

print("Messages sorted successfully.")
print(messages[0])
```

```python
# Calculate response gaps

response_gaps = []

for i in range(1, len(messages)):
    previous_message = messages[i - 1]
    current_message = messages[i]

    # Only calculate when the sender changes
    if previous_message["sender"] != current_message["sender"]:
        gap = current_message["datetime"] - previous_message["datetime"]

        # Ignore gaps longer than 24 hours
        if gap.total_seconds() <= 24 * 60 * 60:
            response_gaps.append(gap.total_seconds() / 60)

if len(response_gaps) > 0:
    average_response = sum(response_gaps) / len(response_gaps)
else:
    average_response = 0

print("Number of response gaps:", len(response_gaps))
print("Average response time:", round(average_response, 2), "minutes")
```

```python
# Find longest silent streak for each participant

all_dates = set()

for msg in messages:
    all_dates.add(msg["date"])

all_dates = sorted(
    all_dates,
    key=lambda x: datetime.strptime(x, "%d/%m/%y")
)

silent_streaks = {}

for person in participants:

    active_dates = set()

    for msg in messages:
        if msg["sender"] == person:
            active_dates.add(msg["date"])

    longest_streak = 0
    current_streak = 0

    for date in all_dates:

        if date not in active_dates:
            current_streak += 1

            if current_streak > longest_streak:
                longest_streak = current_streak

        else:
            current_streak = 0

    silent_streaks[person] = longest_streak

print("Longest silent streaks:")

for person in participants:
    print(person, ":", silent_streaks[person], "days")
```

```python
# Prepare personality statistics

person_stats = {}

for person in participants:
    person_stats[person] = {
        "message_count": 0,
        "total_words": 0,
        "night_messages": 0,
        "uppercase_messages": 0,
        "question_messages": 0,
        "exclamation_messages": 0,
        "humor_messages": 0,
        "care_messages": 0,
        "max_burst": 0
    }

# Keywords
care_keywords = [
    "okay", "safe", "eat", "sleep", "take care",
    "are you", "please", "reminder", "drink water",
    "don't forget"
]

humor_keywords = [
    "lol", "lmao", "haha", "rofl", "lmfao"
]

# Count statistics
for msg in messages:
    person = msg["sender"]
    text = msg["message"]
    lower_text = text.lower()

    person_stats[person]["message_count"] += 1

    words = text.split()
    person_stats[person]["total_words"] += len(words)

    hour = int(msg["time"].split(":")[0])

    # Night messages: 23:00 - 04:59
    if hour >= 23 or hour <= 4:
        person_stats[person]["night_messages"] += 1

    # Uppercase messages
    if len(text.strip()) >= 3 and text.upper() == text:
        person_stats[person]["uppercase_messages"] += 1

    # Questions
    if text.strip().endswith("?"):
        person_stats[person]["question_messages"] += 1

    # Exclamation marks
    if "!!" in text or "!!!" in text:
        person_stats[person]["exclamation_messages"] += 1

    # Humor
    for word in humor_keywords:
        if word in lower_text:
            person_stats[person]["humor_messages"] += 1
            break

    # Caring keywords
    for word in care_keywords:
        if word in lower_text:
            person_stats[person]["care_messages"] += 1
            break

print(person_stats)
```

```python
# Calculate personality percentages

for person in participants:
    stats = person_stats[person]

    message_count = stats["message_count"]

    night_percentage = (
        stats["night_messages"] / message_count
    ) * 100

    uppercase_percentage = (
        stats["uppercase_messages"] / message_count
    ) * 100

    question_percentage = (
        stats["question_messages"] / message_count
    ) * 100

    humor_percentage = (
        stats["humor_messages"] / message_count
    ) * 100

    average_words = (
        stats["total_words"] / message_count
    )

    silent_percentage = (
        silent_streaks[person] / number_of_days
    ) * 100

    print("\n", person)
    print("Messages:", message_count)
    print("Night messages:", round(night_percentage, 2), "%")
    print("Average words/message:", round(average_words, 2))
    print("Uppercase messages:", round(uppercase_percentage, 2), "%")
    print("Questions:", round(question_percentage, 2), "%")
    print("Humor messages:", round(humor_percentage, 2), "%")
    print("Caring messages:", stats["care_messages"])
    print("Silent streak:", silent_streaks[person], "days")
    print("Silent percentage:", round(silent_percentage, 2), "%")
```

```python
# Calculate consecutive message bursts

burst_lengths = {}

for person in participants:
    burst_lengths[person] = []

current_sender = None
current_burst = 0

for msg in messages:
    sender = msg["sender"]

    if sender == current_sender:
        current_burst += 1
    else:
        if current_sender is not None:
            burst_lengths[current_sender].append(current_burst)

        current_sender = sender
        current_burst = 1

# Save the final burst
if current_sender is not None:
    burst_lengths[current_sender].append(current_burst)

print("Average consecutive message burst:")

for person in participants:
    bursts = burst_lengths[person]

    average_burst = sum(bursts) / len(bursts)

    print(person, ":", round(average_burst, 2))
```
